This notebook provides a step-by-step explanation of how AlphaZero works using Tic-Tac-Toe as an example.

## What is AlphaZero?

AlphaZero is a reinforcement learning algorithm that combines:
1. **Monte Carlo Tree Search (MCTS)** - for planning and decision making
2. **Deep Neural Networks** - for position evaluation and move prediction
3. **Self-play** - for generating training data

The algorithm learns to play games without any human knowledge, starting from random play and improving through self-play.
